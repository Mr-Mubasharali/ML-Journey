```python
import pandas as pd
```

```python
df=pd.read_csv('playtennis.csv')
```

```python
df.head()
```

```python
y_n=df['PlayTennis'].value_counts(normalize=True)
```

```python
y_n
```

```python
pd.crosstab(df['Outlook'],df['PlayTennis'],normalize='columns')
```

```python
#To calculate conditional probability use columns
pd.crosstab(df['Outlook'],df['PlayTennis'],normalize='columns').stack()
```

```python
df.columns
```

```python
pro={}
for i in ['Outlook', 'Temperature', 'Humidity', 'Wind']:
    pro.update(pd.crosstab(df[i],df['PlayTennis'],normalize='columns').stack().to_dict())
```

```python
pro
```

```python
pro_y,pro_n=y_n.values
```

```python
pro_y
```

```python
df.head(1)
```

```python
def play_status(outlook,temperature,humidity,wind,pro,pro_y,pro_n):
    yes=pro_y*(pro[outlook,'Yes']*pro[temperature,'Yes']*pro[humidity,'Yes']*pro[wind,'Yes'])
    no=pro_n*(pro[outlook,'No']*pro[temperature,'No']*pro[humidity,'No']*pro[wind,'No'])
    if yes>no:
        print('Yes')
    else:
        print('No')
```

<h4>By changing loc value we can see the result according to the dataset

```python
data=df.iloc[loc].to_list()[:-1]
```

```python
play_status(data[0],data[1],data[2],data[3],pro,pro_y,pro_n)
print(df['PlayTennis'][loc])
```
