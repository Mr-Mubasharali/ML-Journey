<h4> Create a dataset from page source

```python
with open('kit.html','r',encoding='utf-8') as f:
    html=f.read()
```

```python
from bs4 import BeautifulSoup
```

```python
soup=BeautifulSoup(html,'lxml')
```

```python

data=soup.prettify()
```

```python
product_list=soup.find_all('li',class_='product')
```

```python
name=[]
dis_price=[]
org_price=[]
sale_precent=[]
for i in product_list:
    try:
        name.append(i.find_all('a',class_='card-title link-underline card-title-ellipsis')[0].text)
    except:
        name.append(np.nan)
    try:
        dis_price.append(i.find_all('span',class_='price-item price-item--sale')[0].text)
    except:
        dis_price.append(np.nan)
    try:
        org_price.append(i.find_all('s',class_='price-item price-item--regular')[0].text)
    except:
        org_price.append(np.nan)
    try:
        sale_precent.append(i.find_all('span',class_='badge sale-badge')[0].text)
    except:
        sale_precent.append(np.nan)
    
```

```python
import pandas as pd
df=pd.DataFrame({'name':name,'price':org_price,'discount price':dis_price,'sale percentage':sale_precent})
```
