<h4>Apply Ridge Regularization code by self

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
from sklearn.datasets import load_diabetes
```

```python
x,y=load_diabetes(return_X_y=True)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression,ridge_regression
from sklearn.metrics import r2_score,mean_squared_error
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=2)
```

```python
lr=LinearRegression()
```

```python
lr.fit(x_train,y_train)
```

```python
y_pred=lr.predict(x_test)
```

```python
print('Calculate R2 score',r2_score(y_test,y_pred))
print('Calculate RMSE',np.sqrt(mean_squared_error(y_test,y_pred)))
```

<h5> Now Apply Ridge 

```python
from sklearn.linear_model import Ridge
```

```python
R=Ridge(alpha=0.0001)
```

```python
R.fit(x_train,y_train)
```

```python
y_pred1=R.predict(x_test)
```

```python
print('Calculate the R2 score',r2_score(y_test,y_pred1))
print('Calculate the RMSE',np.sqrt(mean_squared_error(y_test,y_pred1)))
```

<h4>Create onw data set and apply Ridge

```python
m=100
x1= 5 * np.random.rand(m,1) -2
x2=0.7 * x1**2 - 2 * x1 + 3 + np.random.randn(m,1)
```

```python
plt.scatter(x1,x2)
```

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import PolynomialFeatures
```

```python
def pred_ridge(x1,x2,alpha):
    model=Pipeline([
        ('ploy_feats',PolynomialFeatures(degree=16)),
        ('ridge',Ridge(alpha=alpha))
    ])
    model.fit(x1,x2)
    return model.predict(x1)
alphas=[0,20,200]
cs=['r','g','b']
plt.figure(figsize=(10,6))
plt.plot(x1,x2,'b+',label='Datapoints')
for alpha, c in zip(alphas,cs):
    preds=pred_ridge(x1,x2,alpha)
    plt.plot(sorted(x1[:,0]),preds[np.argsort(x1[:,0])],c,label='alphas: {}'.format(alpha))
plt.legend()
plt.show
    
```
