<h4>Impute the missing value by Using MICE 

```python
import pandas as pd
import numpy as np
from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer
from sklearn.linear_model import LinearRegression
```

```python
#Custom dataset
df=pd.DataFrame(
    {
        'Age':[25,30,32,40,np.nan],
        'Experience':[np.nan,5,6,7,9],
        'Salary':[40,47,np.nan,60,68],
        'Personal Loan':[1,0,1,1,0]
    }
)
```

```python
df
```

```python
x=df.drop(columns=['Personal Loan'])
y=df['Personal Loan']
```

<h4>On the base of correlation decide the estimate model like linear regression etc

```python
x.corr()
```

```python
model=LinearRegression()
```

```python
mice=IterativeImputer(estimator=model,max_iter=30,imputation_order='roman')
```

```python
r=mice.fit_transform(x)
```
