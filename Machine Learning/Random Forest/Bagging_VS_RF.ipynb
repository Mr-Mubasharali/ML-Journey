<h3> Bagging
<h4> In the bagging on columns base bagging each tree make only the random selected featres like f1,f2 the whole tree make only on f1 and f2 they do not know f3,f4,f5 

<h3>Random Forest
<h4> In the Random forest case on each node select random feature like root note select f1,f2 then the next node select other random feature like f3,f1 etc

```python
import pandas as pd
import numpy as np
from sklearn.datasets import make_classification
import matplotlib.pyplot as plt
```

```python
x,y=make_classification(n_features=5,n_redundant=0,n_informative=5,n_clusters_per_class=1)
```

```python
df=pd.DataFrame(x,columns=['col1','col2','col3','col4','col5'])
df['target']=y
```

```python
df.head()
```

```python
from sklearn.ensemble import BaggingClassifier,RandomForestClassifier
```

```python
bag=BaggingClassifier(max_features=2)
```

```python
bag.fit(df.iloc[:,0:5],df.iloc[:,-1])
```

```python
from sklearn.tree import plot_tree
plt.figure(figsize=(12,9))
plot_tree(bag.estimators_[0],feature_names=df.columns[:-1])
plt.show()
```

```python
rf=RandomForestClassifier(max_features=2)
```

```python
rf.fit(df.iloc[:,0:5],df.iloc[:,-1])
```

```python
plt.figure(figsize=(12,9))
plot_tree(rf.estimators_[0])
plt.show()
```
