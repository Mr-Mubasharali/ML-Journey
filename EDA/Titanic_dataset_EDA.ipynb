# Do some Feature Engineering 

```python
df1=pd.read_csv('train.csv')
```

```python
df=pd.concat([df,df1],ignore_index=True)
```

```python
df['family_size']=df['SibSp']+df['Parch']+1
```

```python
def family_type(num):
    if num==1:
        return 'alon'
    elif num>1 and num <5:
        return 'small'
    else:
        return 'large'
    
```

```python
df['family_type']=df['family_size'].apply(family_type)
```

```python
df['individual_fare']=df['Fare']/df['family_size']
```

```python
df[['Fare','individual_fare']].describe()
```

```python
pd.crosstab(df['Survived'],df['family_type'],normalize='columns')*100
```

```python
df['surname']=df['Name'].str.split(',').str.get(0)
```

```python
df['title']=df['Name'].str.split(',').str.get(1).str.get(0).str.strip()
```

```python
rare_titles = ['Dr', 'Rev', 'Col', 'Major', 'Lady', 'Sir', 'Capt']

df['title'] = df['title'].replace(rare_titles, 'rare')

```

```python
df['title'].value_counts()
```

```python
df[df['title']=='others']
```

```python
df['Cabin'].fillna('M',inplace=True)
df['Cabin'] = df['Cabin'].fillna('M')
```

```python
df[df['Cabin']=='M']
```

```python
df['deck']=df['Cabin'].str.get(0)
```

```python
pd.crosstab(df['deck'],df['Survived'],normalize='index').plot(kind='bar',stacked=True)
```

```python
df.drop(columns='Survived',axis=0,inplace=True)
```

```python
df['Fare'].dropna(inplace=True)
```
