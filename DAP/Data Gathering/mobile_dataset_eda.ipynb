```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv('Mobile_datase.csv')
```

```python
df1=df.copy()
```

```python
df1[df1['Phone_price'].isna()]
```

```python
df1.drop('Unnamed: 0',axis=1,inplace=True)
```

```python
df1['Phone_price']=df1['Phone_price'].astype('float')
```

```python
df1['Discounted_price']=df1['Discounted_price'].astype('float')
```

```python
df1.reset_index(inplace=True)
```

```python
temp_df=df1[df1['Phone_price'].isna()]
```

```python
temp_df['Phone_price']=temp_df['Discounted_price'].values
```

```python
x_index=temp_df.index
x_values=temp_df.iloc[:,4:6].values
```

```python
df1.iloc[x_index,4:6]=x_values
```
